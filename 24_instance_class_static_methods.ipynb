```python
#Instance method
class mentor:
    def __init__(self,name):
        self.name = name

    def introduction(self):
        return f"I am a mentor {self.name} on euron"
   
```

```python
a = mentor("prince")
```

```python
a.introduction()
```

```python
b = mentor("Francis")
```

```python
b.introduction()
```

```python
#class method
class course:
    total_course = 0

    def __init__(self,title):
        self.title = title
        course.total_course =course.total_course+1

    @classmethod
    def get_total_course(cls):
        return cls.total_course
```

```python
a = course("python")
```

```python
b = course("ML ops")
```

```python
b.get_total_course()
```

```python
a.get_total_course()
```

```python
course.get_total_course()
```

```python
#Static method
class students:

    @staticmethod
    def is_valid_email(email):
        return "@" in email and "." in email
```

```python
students.is_valid_email("prince.francis64@gmail.com")
```
