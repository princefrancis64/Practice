```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .master("local[*]") \
    .appName("FileWordCount") \
    .getOrCreate()

sc = spark.sparkContext
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
rdd_1 = sc.textFile("sample (1).txt")

print(rdd_1.collect())
```

```python
rdd_2 = rdd_1.flatMap(lambda line: line.split(" "))
print(rdd_2.collect())
```

```python
rdd_3 = rdd_2.map(lambda word: (word,1))
print(rdd_3.collect())
```

```python
rdd_4 = rdd_3.reduceByKey(lambda x,y : x+y)
print(rdd_4.collect())
```
