```python
base_rdd = spark.sparkContext.textFile("orders.txt")
```

```python
base_rdd.take(5)
```

```python
mapped_rdd = base_rdd.map(lambda x: (x.split(",")[3],x.split(",")[2]))
```

```python
mapped_rdd.take(5)
```

```python
grouped_rdd = mapped_rdd.groupByKey()
```

```python
result = grouped_rdd.map(lambda x: (x[0],len(x[1])))
```
