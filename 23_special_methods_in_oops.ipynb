```python
class course:
    def __init__(self,title,mentor):
        self.title = title
        self.mentor = mentor

    def __str__(self):
        return f"My name is {self.mentor}"
    

    def __eq__(self,value):
        return self.title ==  value.title

    def __len__(self):
        return self.title
    
    def __del__(self):
        print("delete the object")

    def __call__(self,*args,**kwargs):
        return f"{self.title},{self.mentor},{args},{kwargs}"
```

```python
a = course("python","prince")
```

```python
print(a)
```

```python
a.mentor
```
