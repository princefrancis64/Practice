```python
def decorator_fun(func):
    def wrapper():
        print("this will be printed before the function")
        func()
        print("this will be printed after the function")
    return wrapper
```

```python
@decorator_fun
def say_hi():
    print("hi euron")
```

```python
say_hi()
```

```python
import time

def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        print("time taken by this function is ",time.time()-start)
        return result
    return wrapper
```

```python
def listparser(l):
    print(l)
    for i in l:
        print(i)
```

```python
listparser([1,2,3,4,4,5,5,6,6])
```

```python
@timer
def listparser(l):
    print(l)
    for i in l:
        print(i)
```

```python
listparser([1,2,3,4,4,5,5,6,6])
```

```python
def decorator_fun1(func):
    def wrapper(*args,**kwargs):
        print("this will be printed before the function")
        func(*args,**kwargs)
        print("this will be printed after the function")
    return wrapper
```

```python
@timer
@decorator_fun1
def listparser(l):
    print(l)
    for i in l:
        print(i)
```

```python
listparser([1,2,3,4,4,5,5,6,6])
```
