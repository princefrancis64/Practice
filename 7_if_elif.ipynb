```python
gpa = 3.7
test_score = 85
extra_curicular = True

if gpa > 3.5:
    if test_score > 90:
        print("You will get a scholarship")
    elif test_score > 80:
        if extra_curicular:
            print("You will get a partial scholarship")
        else:
            print("You will get admission but no scholarship")
    else:
        print("You have some chance to get an admission")
elif gpa > 3.0:
    if test_score>85:
        print("You will get admission")
    else:
        print("you will get admission after some review")
else:
    print("you won't get admission")
```

```python
#match case

day = "Monday"
match day:
    case "Monday":
        print("this is start of the week")
    case "tuesday":
        print("this is a second day of the week")
    case "wednesday":
        print("this is mid of the week")
    case _:
        print("its just another day")
```
